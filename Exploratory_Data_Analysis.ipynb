## 3. ***Data Wrangling***

### Data Wrangling Code

````python

# DATA WRANGLING

# 1. Remove duplicate records
titles = titles.drop_duplicates().copy()
credits = credits.drop_duplicates().copy()

# 2. Handle missing values carefully
# Preserve missing values for analysis instead of replacing them blindly

# 3. Convert genres into list format
import ast

def clean_genres(value):
    if isinstance(value, list):
        return value
    if pd.isna(value):
        return []
    try:
        return ast.literal_eval(value)
    except (ValueError, SyntaxError):
        return []

titles['genres'] = titles['genres'].apply(clean_genres)

# 4. Correct data types
titles['release_year'] = pd.to_numeric(
    titles['release_year'], errors='coerce'
)

titles['runtime'] = pd.to_numeric(
    titles['runtime'], errors='coerce'
)

# 5. Group credits using common ID
credits_summary = credits.groupby('id').agg({
    'name': lambda x: ', '.join(x.dropna().astype(str).unique()),
    'role': lambda x: ', '.join(x.dropna().astype(str).unique())
}).reset_index()

# 6. Merge both datasets
netflix = pd.merge(
    titles,
    credits_summary,
    on='id',
    how='left'
)

# 7. Display final dataset
print("Data Wrangling Completed Successfully!")
print("Final Dataset Shape:", netflix.shape)

display(netflix.head())
````

#### Chart - 14 - Correlation Heatmap

````python
# Select numerical columns
numeric_data = titles.select_dtypes(include=['number'])

# Calculate correlation
correlation = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Netflix Dataset', fontsize=15)
plt.tight_layout()
plt.show()
````

#### Chart - 15 - Pair Plot

````python
# Select important numerical columns
pair_data = titles[
    ['release_year', 'runtime', 'imdb_score',
     'imdb_votes', 'tmdb_popularity', 'tmdb_score']
].dropna()

# Take a sample to make the plot easier to display
pair_data = pair_data.sample(
    n=min(500, len(pair_data)),
    random_state=42
)

# Create pair plot
sns.pairplot(pair_data, diag_kind='hist')

plt.suptitle('Pair Plot of Netflix Numerical Variables', y=1.02)
plt.show()
````

##### 1. Why did you pick the specific chart?

I selected a pair plot because it helps visualize relationships between multiple numerical variables at the same time. It displays scatter plots between variables and histograms showing their individual distributions.

##### 2. What is/are the insight(s) found from the chart?

The pair plot helps us identify patterns, trends, and possible relationships between variables such as IMDb score, IMDb votes, TMDB popularity, runtime, and release year. It also helps identify clusters and unusual data points that may need further investigation.
